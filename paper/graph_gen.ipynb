This is a notebook which generates the graph which appear in the paper *Nash Stability in Fractional Hedonic Games with Bounded Size Coalitions*.

```python
import networkx as nx
import numpy as np
import itertools as ite

from pyhedonic import hedonicgame as hg
```

```python
def draw_graph(G: hg.HedonicGame, filename: str) -> None:
    """
    Draws a graph using networkx and saves it to a file.
    """
    dotG = G.graph.to_dot()
    for node in dotG.get_nodes():
        node.set_shape("circle")
        node.set_penwidth("2.0")
    for edge in dotG.get_edges():
        label = edge.get_label()
        if label is not None:
            # edge.set_label("\u202F"+ str(label))
            edge.set_label(f"""<
                <TABLE BORDER="0" CELLBORDER="0" CELLSPACING="0" CELLPADDING="0">
                <TR>
                <TD WIDTH="3" FIXEDSIZE="TRUE"></TD>
                <TD>{label}</TD>
                </TR>
            </TABLE>
            >""")
        edge.set_penwidth("2.0")
    dotG.set_edge_defaults(penwidth="2.5")
    dotG.write_pdf(filename)
```

```python
draw_graph(hg.GAME_K3_NOEQUILIBRIUM, "GAME_K3_NOEQUILIBRIUM.pdf")
draw_graph(hg.GAME_K7_NOEQUILIBRIUM_SIMPLE, "GAME_K7_NOEQUILIBRIUM_SIMPLE.pdf")
draw_graph(hg.GAME_K3_CYCLIC_DYNAMIC, "GAME_K3_CYCLIC_DYNAMIC.pdf")
```

```python
G = hg.GAME_K3_CYCLIC_DYNAMIC.coalition_structures_as_nx()
Gbest = hg.GAME_K3_CYCLIC_DYNAMIC.coalition_structures_as_nx(best_response_only=True)
for g in G.nodes:
    if G.out_degree(g) == 0:
        G.nodes[g]["style"] = "filled"
        G.nodes[g]["fillcolor"] = "green"
for e in G.edges:
    ag = G.edges[e]["agent"]
    co = G.edges[e]["target_coalition"]
    v = e[0].deviation_value(ag, co)
    G.edges[e]["label"] = f"{ag} → {e[0][co] or '{}'}\n{v[0]:.2f}"
for g in Gbest.edges:
    G.edges[g]["penwidth"] = 3
for cs in hg.GAME_K3_CYCLIC_DYNAMIC_CYCLE:
    G.nodes[cs]["style"] = "filled"
    G.nodes[cs]["fillcolor"] = "orange"
for cs1, cs2 in ite.pairwise(
    hg.GAME_K3_CYCLIC_DYNAMIC_CYCLE + [hg.GAME_K3_CYCLIC_DYNAMIC_CYCLE[0]]
):
    G.edges[cs1, cs2]["color"] = "red"
    G.edges[cs1, cs2]["fontcolor"] = "red"
Hdot = nx.nx_pydot.to_pydot(G)
Hdot.write_pdf("GAME_K3_CYCLIC_DYNAMIC_FULLGRAPH.pdf")
```

```python
G = hg.GAME_K3_CYCLIC_DYNAMIC.coalition_structures_as_nx(best_response_only=True)
for g in G.edges:
    G.edges[g]["label"] = ""
for cs in hg.GAME_K3_CYCLIC_DYNAMIC_CYCLE:
    G.nodes[cs]["style"] = "filled"
    G.nodes[cs]["fillcolor"] = "lightgreen"
for cs1, cs2 in ite.pairwise(
    hg.GAME_K3_CYCLIC_DYNAMIC_CYCLE + [hg.GAME_K3_CYCLIC_DYNAMIC_CYCLE[0]]
):
    G.edges[cs1, cs2]["color"] = "red"
    G.edges[cs1, cs2]["penwidth"] = 3
Hdot = nx.nx_pydot.to_pydot(G)
Hdot.write_pdf("GAME_K3_CYCLIC_DYNAMIC_BESTGRAPH.pdf")
```

```python
hg.GAME_K3_DISCONNECTED_DYNAMIC.graph.to_dot().write_pdf("GAME_K3_DISCONNECTED_DYNAMIC.pdf")
```

```python
G = hg.GAME_K3_DISCONNECTED_DYNAMIC.coalition_structures_as_nx()
u = hg.CoalitionStructure(hg.GAME_K3_DISCONNECTED_DYNAMIC, np.array([0, 1, 1, 1]))
reachable = nx.descendants(G, u) | {u}
H = G.subgraph(reachable).copy()
Hdot = nx.nx_pydot.to_pydot(H)
Hdot.set_layout("circo")
Hdot.write_pdf("GAME_K3_DISCONNECTED_DYNAMIC_CYCLE.pdf")
```
